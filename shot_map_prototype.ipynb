# Shot map prototype: 2022 World Cup Final
First experiment before the Streamlit app: pull one match from StatsBomb open data (`match_id=3869685`, Argentina vs France) and plot every shot, sized by xG.

- Open data needs no credentials. The `NoAuthWarning` is expected.
- Penalties are dropped here to look at open play only.
- StatsBomb coordinates always point towards the right goal, so France's shots are mirrored (`120-x`, `80-y`) to put both teams on one pitch.
- White = goal, red = no goal, bubble size = xG.


```python
import pandas as pd
from statsbombpy import sb
```

```python
df=sb.events(match_id=3869685)
```

```python
df.columns
```

```python
df = df[df['type'] == 'Shot']
df=df[['team','player','location','shot_outcome','shot_statsbomb_xg','shot_type']]
df.head()
```

```python
df=df[df['shot_type'] != 'Penalty']
df.head()
```

```python
df[['x','y']] = df['location'].apply(pd.Series)
df.head()
```

```python
from mplsoccer import Pitch
pitch=Pitch(pitch_type='statsbomb')
pitch = Pitch(
    pitch_color='grass',
    line_color='white', 
    stripe=True, 
    linewidth=3
)
fig,ax = pitch.draw()
for x in df.to_dict(orient='records'):
    if x['team']=='France':
        pitch.scatter(x=120-x['x'],y=80-x['y'],ax=ax,s=500*x['shot_statsbomb_xg'],ec='black',c='white' if x['shot_outcome']=='Goal' else 'red')
    else:
        pitch.scatter(x=x['x'],y=x['y'],ax=ax,s=500*x['shot_statsbomb_xg'],ec='black',c='white' if x['shot_outcome']=='Goal' else 'red')
```
